```python
alumnos = [
    ['nadine', 1975, 'f', 'Reins'],
    ['raquel', 1970, 'f', 'Oviedo'],
    ['sergio', 1960, 'm', 'gijón'],
    ['alvaro', 1990, 'm', 'León'],
    ['yeray', 2003, 'm', 'miranda de ebro'],
]

asturianos = []
moces = []
mozos = []
jovenzuelos = []

alumnos.append(['raquel', 1970, 'f', 'Oviedo'])
alumnos.append(['sergio', 1960, 'm', 'gijón'])

for alumno in alumnos:
    nombre, anio_nacimiento, genero, lugar_origen = alumno
    print(nombre)

    # Example of how to populate 'asturianos' based on location
    if lugar_origen in ['Oviedo', 'gijón']:
        asturianos.append(alumno)

    # Example of how to populate 'moces', 'mozos', 'jovenzuelos' based on age
    # Assuming 'moces' for younger females, 'mozos' for younger males, 'jovenzuelos' for very young
    if genero == 'f' and anio_nacimiento > 1980:
        moces.append(alumno)
    elif genero == 'm' and anio_nacimiento > 1980:
        mozos.append(alumno)
    if anio_nacimiento > 2000: # Just an example condition
        jovenzuelos.append(alumno)

print('\nAlumnos asturianos:')
for alumno in asturianos:
    print(alumno[0])

print('\nAlumnas jovenes (moces):')
for alumno in moces:
    print(alumno[0])

print('\nAlumnos jovenes (mozos):')
for alumno in mozos:
    print(alumno[0])

print('\nAlumnos muy jovenes (jovenzuelos):')
for alumno in jovenzuelos:
    print(alumno[0])
```
