```python
# calcular la suma de:

def sumar(a,b):   # definimos nombre y parametros de la función
  print(f'la suma de {a} + {b} es {a+b}')   # definimos lo que hara la funcion

#llamamos varias veces a la funcion pasandole parametros diferetes:
sumar(5,3)
sumar(6,4)
sumar(7,5)
```

```python
def saludar(nombre, momento_del_dia):
  match momento_del_dia.lower():
    case 'mañana':
      print(f'¡Buenos días, {nombre}!')
    case 'tarde':
      print(f'¡Buenas tardes, {nombre}!')
    case 'noche':
      print(f'¡Buenas noches, {nombre}!')
    case _:
      print(f'¡Hola {nombre}, que tengas un buen {momento_del_dia}!')

saludar('Yeray', 'mañana')
saludar('María', 'tarde')
saludar('Juan', 'noche')
saludar('Ana', 'día')
```
