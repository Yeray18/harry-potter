```python
# cargar libreria de html
from IPython.display import HTML

# defino la variable donde almacenare todo el codigo html
mihtml= """<h1> soy un titulo de nivel 1</h1>"""     # h1  = titulo 1
mihtml+= """<h2> yo un subtitulo o titulo de nivel 2</h2>"""     # h2  = titulo 2
mihtml+= """<h3> titulo de nivel 3</h3>"""     # h3  = titulo 3... etc h6 (corregido </p> a </h3>)
mihtml+= """<p> este es un parrafo</p>"""     # p  = parrafo

# Las siguientes líneas reasignaban mihtml, perdiendo el contenido anterior.
# He cambiado = por += para concatenar y he corregido las etiquetas HTML.
mihtml+= """<ul>""" # Abre una lista no ordenada
mihtml+= """<li> soy un perro de lista 1</li>"""
mihtml+= """<li> soy un zorro de lista 2</li>"""
mihtml+= """<li> soy un tortuga de lista 3</li>"""
mihtml+= """<li> soy un avestruz de lista 4</li>"""
mihtml+= """<li> soy un delfin de lista 5</li>"""
mihtml+= """</ul>""" # Cierra la lista no ordenada

# La línea original 'mihtml=+= """<' tenía dos errores:
# 1. 'mihtml=+= ' no es una sintaxis válida (debe ser 'mihtml +=')
# 2. La cadena """<""" no estaba cerrada.
# Si se deseaba añadir más HTML, se haría así:
mihtml+= """<br>Esto es un texto adicional después de la lista."""

# lanzo codigo html
display(HTML(mihtml)) # Corregido 'html' a 'HTML'
```
